NumPy can read and write text-delimited data and its own binary formats. Use `.npy` for one array and `.npz` for multiple named arrays.

```python
import numpy as np
from pathlib import Path

DATA_DIR = Path('../data')
DATA_DIR.mkdir(parents=True, exist_ok=True)
arr = np.arange(12, dtype=np.float64).reshape(3, 4)
np.save(DATA_DIR / 'example.npy', arr)
loaded = np.load(DATA_DIR / 'example.npy')
print('loaded:\n', loaded)
np.savez_compressed(DATA_DIR / 'arrays.npz', features=arr, targets=arr.sum(axis=1))
with np.load(DATA_DIR / 'arrays.npz') as archive:
    print('keys:', archive.files)
    print('targets:', archive['targets'])
```

```python
# CSV-style text export; fmt and delimiter make the format explicit.
csv_path = DATA_DIR / 'example.csv'
np.savetxt(csv_path, arr, delimiter=',', header='c1,c2,c3,c4', comments='')
text_loaded = np.loadtxt(csv_path, delimiter=',', skiprows=1)
print('text round-trip successful:', np.allclose(arr, text_loaded))
```
