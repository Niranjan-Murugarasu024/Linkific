Vectorization expresses work as array operations instead of iterating through individual values in Python. Benchmark both methods on the same data and verify they produce the same result.

```python
import numpy as np
from time import perf_counter

rng = np.random.default_rng(123)
x = rng.random(1_000_000)

def python_loop(values):
    result = np.empty_like(values)
    for i, value in enumerate(values):
        result[i] = value * value + 2 * value + 1
    return result

def vectorized(values):
    return values * values + 2 * values + 1

# Warm-up
expected = vectorized(x)
start = perf_counter(); loop_result = python_loop(x); loop_seconds = perf_counter() - start
start = perf_counter(); vector_result = vectorized(x); vector_seconds = perf_counter() - start
print(f'loop: {loop_seconds:.4f} s')
print(f'vectorized: {vector_seconds:.4f} s')
print(f'speedup: {loop_seconds / vector_seconds:.2f}x')
print('results match:', np.allclose(loop_result, vector_result))
```

## Benchmarking guidance
- Run multiple trials and compare the median time.
- Avoid timing notebook display or data creation as part of the operation.
- Results depend on hardware, NumPy build, array size, memory bandwidth, and system load.
- Vectorized expressions may create temporary arrays; fewer Python loops does not guarantee the lowest memory footprint.
