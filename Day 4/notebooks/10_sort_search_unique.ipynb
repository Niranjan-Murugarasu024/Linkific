Sorting and searching are common preprocessing operations. Remember that `argsort` returns indices, not sorted values.

```python
import numpy as np

x = np.array([7, 2, 9, 2, 5, 7, 1])
print('sorted:', np.sort(x))
print('sort indices:', np.argsort(x))
print('unique:', np.unique(x))
values, counts = np.unique(x, return_counts=True)
print('frequency table:', dict(zip(values.tolist(), counts.tolist())))
```

```python
sorted_x = np.sort(x)
print('binary search insertion position for 5:', np.searchsorted(sorted_x, 5))
print('index of maximum:', np.argmax(x))
print('maximum value:', x[np.argmax(x)])
print('indices of values >= 7:', np.where(x >= 7))
```
