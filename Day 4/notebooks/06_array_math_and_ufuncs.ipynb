NumPy arithmetic is element-wise by default. Universal functions (ufuncs) operate efficiently over arrays.

```python
import numpy as np

a = np.array([1, 2, 3, 4], dtype=float)
b = np.array([10, 20, 30, 40], dtype=float)
print('a + b:', a + b)
print('b - a:', b - a)
print('a * b:', a * b)
print('b / a:', b / a)
print('a squared:', a ** 2)
print('square root:', np.sqrt(a))
print('exponential:', np.exp(a))
print('log:', np.log(a))
```

```python
angles = np.array([0, np.pi / 2, np.pi])
print('sin:', np.sin(angles))
print('rounded:', np.round(np.array([1.234, 2.678]), 2))
print('absolute:', np.abs(np.array([-3, -1, 2])))
print('clipped:', np.clip(np.array([-5, 0, 12]), 0, 10))
```
