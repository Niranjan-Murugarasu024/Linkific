```python
import numpy as np

zeros = np.zeros((2, 3))
ones = np.ones((2, 3), dtype=np.int64)
full = np.full((2, 3), 7)
identity = np.eye(3)
empty = np.empty((2, 2))  # Values are uninitialized: overwrite before reading.

print('zeros:\n', zeros)
print('ones:\n', ones)
print('full:\n', full)
print('identity:\n', identity)
print('empty shape:', empty.shape)
```

```python
# arange uses a step; the stop value is excluded.
sequence = np.arange(0, 10, 2)
# linspace creates a specified number of evenly spaced values, including endpoints by default.
points = np.linspace(0, 1, 5)
print('arange:', sequence)
print('linspace:', points)
print('integer range:', np.arange(5, 10))
```
