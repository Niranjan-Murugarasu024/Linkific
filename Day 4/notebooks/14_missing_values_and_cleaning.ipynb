NumPy commonly represents missing numeric values as `np.nan`. Build explicit masks before calculations.

```python
import numpy as np

temperatures = np.array([21.5, 22.1, np.nan, 23.0, -999, 22.7, np.nan])
missing_mask = np.isnan(temperatures)
sentinel_mask = temperatures == -999
valid_mask = (~missing_mask) & (~sentinel_mask)
clean = temperatures[valid_mask]
print('missing count:', missing_mask.sum())
print('sentinel count:', sentinel_mask.sum())
print('clean values:', clean)
print('clean mean:', clean.mean())
```

```python
# Example imputation: replace NaNs with the mean of valid values.
values = np.array([1.0, 2.0, np.nan, 4.0, 5.0])
mean_value = np.nanmean(values)
print(mean_value)
imputed = np.where(np.isnan(values), mean_value, values)
print('mean used:', mean_value)
print('imputed:', imputed)

# Keep the original data unchanged unless you intentionally assign into it.
```

## Caution
A sentinel such as `-999` is domain-specific; do not treat it as missing unless the dataset documentation says so. For integer arrays, NaN requires converting to a floating dtype or using a different missing-data representation.
