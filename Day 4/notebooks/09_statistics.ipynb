Explore central tendency, spread, quantiles, and missing-data-aware statistics.

```python
import numpy as np

data = np.array([12, 15, 15, 18, 21, 24, 30], dtype=float)
print('mean:', np.mean(data))
print('median:', np.median(data))
print('population std:', np.std(data))
print('sample std:', np.std(data, ddof=1))
print('variance:', np.var(data))
print('percentiles:', np.percentile(data, [25, 50, 75, 90]))
print('min/max/range:', data.min(), data.max(), np.ptp(data))
```

```python
measurements = np.array([10.0, 12.0, np.nan, 15.0, 18.0])
print('ordinary mean:', np.mean(measurements))
print('NaN-aware mean:', np.nanmean(measurements))
print('NaN-aware median:', np.nanmedian(measurements))
print('NaN-aware standard deviation:', np.nanstd(measurements))
```

## Interpretation
`ddof=0` calculates population standard deviation; `ddof=1` gives the common sample standard-deviation estimator. Choose based on what your data represents.
