Use the modern `np.random.default_rng` API. A fixed seed makes examples reproducible.

```python
import numpy as np

rng = np.random.default_rng(seed=42)
print('integers:', rng.integers(0, 10, size=8))
print('uniform:', rng.random(3))
print('normal:', rng.normal(loc=100, scale=15, size=5))
print('choice:', rng.choice(['A', 'B', 'C'], size=6, replace=True))
```

```python
rng = np.random.default_rng(7)
measurements = rng.normal(loc=50, scale=8, size=1000)
print('count:', measurements.size)
print('sample mean:', measurements.mean())
print('sample standard deviation:', measurements.std(ddof=1))
print('percentiles:', np.percentile(measurements, [5, 50, 95]))
```
