## Objectives
- Import NumPy using the conventional alias `np`.
- Inspect the installed version.
- Understand the difference between a Python list and a NumPy array.

```python
import numpy as np

print('NumPy version:', np.__version__)
python_list = [1, 2, 3, 4]
array = np.array(python_list)
print('List:', python_list, type(python_list))
print('Array:', array, type(array))
print('Array dtype:', array.dtype)
print('Array shape:', array.shape)
print('Number of dimensions:', array.ndim)
print('Number of elements:', array.size)
```

```python
# A NumPy array usually stores elements of one common dtype.
values = np.array([1, 2, 3, 4], dtype=np.float64)
print(values)
print('dtype:', values.dtype)
print('bytes per element:', values.itemsize)
print('total bytes:', values.nbytes)
```
