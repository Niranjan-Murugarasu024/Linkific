NumPy supports dot products, matrix multiplication, norms, and solving linear systems.

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print('dot product:', np.dot(a, b))
print('norm:', np.linalg.norm(a))
A = np.array([[1., 2.], [3., 4.]])
B = np.array([[5., 6.], [7., 8.]])
print('element-wise product:\n', A * B)
print('matrix product:\n', A @ B)
```

```python
# Solve A x = b. Prefer solve() to explicitly calculating inv(A) @ b.
A = np.array([[3., 1.], [1., 2.]])
b = np.array([9., 8.])
x = np.linalg.solve(A, b)
print('solution:', x)
print('check A @ x:', A @ x)
print('determinant:', np.linalg.det(A))
print('eigenvalues:', np.linalg.eigvals(A))
```
