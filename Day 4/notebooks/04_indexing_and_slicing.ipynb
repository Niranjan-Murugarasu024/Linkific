Indexing selects individual values; slicing selects ranges. The slice stop index is excluded.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])
print('first:', arr[0])
print('last:', arr[-1])
print('index 1 through 3:', arr[1:4])
print('every second value:', arr[::2])
print('reversed:', arr[::-1])
```

```python
matrix = np.arange(1, 13).reshape(3, 4)
print(matrix)
print('row 0:', matrix[0])
print('element row 1 col 2:', matrix[1, 2])
print('first two rows:\n', matrix[:2, :])
print('last column:', matrix[:, -1])
print('submatrix:\n', matrix[1:, 1:3])
```

## Important note
Basic slices are often views into the original array. Changing a view can change the original array.

```python
original = np.arange(6)
view = original[1:4]
view[0] = 99
print('view:', view)
print('original changed:', original)
copy = original[1:4].copy()
copy[0] = -5
print('copy changed:', copy)
print('original unchanged by copy:', original)
```
