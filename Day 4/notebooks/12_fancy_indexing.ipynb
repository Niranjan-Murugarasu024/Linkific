Fancy indexing selects elements using integer index arrays. Unlike basic slicing, advanced indexing generally returns a copy.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
indices = np.array([0, 2, 4])
print('selected:', arr[indices])
arr[[1, 3]] = -1
print('after indexed assignment:', arr)
```

```python
matrix = np.arange(1, 17).reshape(4, 4)
print(matrix)
rows = np.array([0, 2, 3])
cols = np.array([1, 2, 0])
print('paired coordinates:', matrix[rows, cols])
print('selected rows:\n', matrix[[0, 3]])
selected_copy = matrix[[0, 3]]
selected_copy[0, 0] = 999
print('original remains:', matrix[0, 0])
print('Changes Copy:\n', selected_copy)
```
