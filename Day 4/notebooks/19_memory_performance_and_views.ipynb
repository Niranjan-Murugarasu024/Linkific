Performance includes both execution time and memory traffic. Learn when operations allocate arrays.

```python
import numpy as np

x = np.arange(1_000_000, dtype=np.float64)
view = x[100:200]
copy = x[100:200].copy()
print('x bytes:', x.nbytes)
print('view shares memory:', np.shares_memory(x, view))
print('copy shares memory:', np.shares_memory(x, copy))
```

```python
a = np.arange(1_000_000, dtype=np.float64)
# Creates a new result array:
b = a * 2
# In-place multiplication reuses a's allocated storage:
a *= 2
print('equal:', np.array_equal(a, b))

# `out=` can reuse an existing output buffer for many ufuncs.
source = np.arange(5.0)
out = np.empty_like(source)
np.multiply(source, 3, out=out)
print('out buffer:', out)
```

## Practical rules
- Use an appropriate dtype; do not downcast without checking precision/range requirements.
- Avoid unnecessary copies of very large arrays.
- Prefer vectorized operations, but profile before optimizing.
- Use `np.shares_memory` to investigate view/copy behavior.
