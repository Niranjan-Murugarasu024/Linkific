Reshape changes the dimensions without changing element order. The requested shape must have the same number of elements.

```python
import numpy as np

arr = np.arange(12)
print('original:', arr)
print('3x4:\n', arr.reshape(3, 4))
print('2x2x3 shape:\n', arr.reshape(2, 2, 3))
print('infer one dimension:', arr.reshape(3, -1).shape)
```

```python
matrix = np.arange(1, 7).reshape(2, 3)
print('matrix:\n', matrix)
print('transpose:\n', matrix.T)
print('flatten copy:', matrix.flatten())
print('ravel view when possible:', matrix.ravel())
print('add axis:', matrix[:, np.newaxis, :].shape)
print('remove singleton axis:', matrix[:, np.newaxis, :].squeeze())
```
