Aggregations reduce values, for example with `sum`, `mean`, `min`, and `max`. For a 2D array, `axis=0` reduces rows and returns one value per column; `axis=1` reduces columns and returns one value per row.

```python
import numpy as np

x = np.array([[1, 2, 3], [4, 5, 6]])
print('sum all:', np.sum(x))
print('sum axis 0 (column totals):', np.sum(x, axis=0))
print('sum axis 1 (row totals):', np.sum(x, axis=1))
print('mean columns:', np.mean(x, axis=0))
print('min rows:', np.min(x, axis=1))
print('max:', np.max(x))
```

```python
scores = np.array([[78, 85, 90], [92, 88, 84], [65, 70, 72]])
subjects = ['Math', 'Science', 'Computing']
students = ['Asha', 'Ben', 'Chen']
print('student averages:', dict(zip(students, scores.mean(axis=1))))
print('subject averages:', dict(zip(subjects, scores.mean(axis=0))))
```
