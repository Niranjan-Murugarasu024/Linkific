Learn `shape`, `ndim`, `size`, `dtype`, `itemsize`, and `nbytes`.

```python
import numpy as np

matrix = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int32)
for name in ('shape', 'ndim', 'size', 'dtype', 'itemsize', 'nbytes'):
    print(f'{name}: {getattr(matrix, name)}')
```

```python
integers = np.array([1, 2, 3], dtype=np.int16)
floats = integers.astype(np.float32)
print('before:', integers, integers.dtype)
print('after: ', floats, floats.dtype)
print('int16 bytes:', integers.nbytes, '| float32 bytes:', floats.nbytes)

# Be careful: converting a floating-point value to an integer discards its fractional part.
print(np.array([2.9, 3.1]).astype(np.int64))
```
