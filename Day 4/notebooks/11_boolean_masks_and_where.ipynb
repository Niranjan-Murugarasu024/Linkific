Boolean masks select values based on a condition. Combine multiple conditions with `&`, `|`, and `~`; wrap each comparison in parentheses.

```python
import numpy as np

scores = np.array([45, 67, 82, 39, 91, 74, 58])
passed = scores >= 50
print('mask:', passed)
print('passing scores:', scores[passed])
print('scores from 60 through 85:', scores[(scores >= 60) & (scores <= 85)])
print('failed scores:', scores[scores < 50])
```

```python
values = np.array([-3, -1, 0, 2, 5])
labels = np.where(values >= 0, 'non-negative', 'negative')
clipped = np.where(values < 0, 0, values)
print('labels:', labels)
print('replace negatives with zero:', clipped)
print('count positives:', np.count_nonzero(values > 0))
```
