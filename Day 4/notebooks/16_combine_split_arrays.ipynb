Choose concatenation when joining along an existing axis and stacking when introducing a new axis.

```python
import numpy as np

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print('concatenate:', np.concatenate([a, b]))
print('stack axis 0:\n', np.stack([a, b], axis=0))
print('stack axis 1:\n', np.stack([a, b], axis=1))
print('vertical stack:\n', np.vstack([a, b]))
print('horizontal stack:', np.hstack([a, b]))
```

```python
matrix = np.arange(12).reshape(3, 4)
print(matrix)
left, right = np.split(matrix, 2, axis=1)
print('left half:\n', left)
print('right half:\n', right)
parts = np.array_split(np.arange(10), 3)
print('unequal splits:', parts)
```
