Broadcasting lets NumPy combine compatible shapes without manually copying values. Dimensions are compatible when equal or when one is 1.

```python
import numpy as np

matrix = np.array([[1, 2, 3], [4, 5, 6]])
row = np.array([10, 20, 30])
print('matrix + row:\n', matrix + row)
column = np.array([[100], [200]])
print('matrix + column:\n', matrix + column)
print('shape (2, 3) + (3,):', (matrix + row).shape)
```

```python
# Normalize each column by subtracting its column mean.
data = np.array([[10., 20., 30.], [20., 30., 40.], [30., 40., 50.]])
column_means = data.mean(axis=0)
centered = data - column_means
print('column means:', column_means)
print('centered:\n', centered)
print('new column means:', centered.mean(axis=0))
```
