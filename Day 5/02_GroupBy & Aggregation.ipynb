```python
import pandas as pd
import numpy as np

sales = pd.DataFrame({
    "region": [
        "South", "North", "South", "East", "West",
        "North", "East", "West", "South", "North"
    ],
    "product": [
        "Laptop", "Phone", "Phone", "Laptop", "Tablet",
        "Laptop", "Phone", "Tablet", "Laptop", "Phone"
    ],
    "quantity": [2, 5, 3, 1, 4, 3, 8, 2, 1, 6],
    "unit_price": [50000, 20000, 20000, 50000, 15000,
                   50000, 20000, 15000, 50000, 20000],
    "discount": [0.05, 0.10, 0.00, 0.05, 0.10,
                 0.00, 0.05, 0.00, 0.10, 0.05]
})

sales["revenue"] = (
    sales["quantity"] * sales["unit_price"] * (1 - sales["discount"])
)

display(sales)
```

```python
print("Total revenue:", sales["revenue"].sum())
print("Average revenue:", sales["revenue"].mean())
print("Maximum revenue:", sales["revenue"].max())
print("Minimum revenue:", sales["revenue"].min())
```

```python
region_summary = sales.groupby("region").agg(
    total_revenue=("revenue", "sum"),
    average_revenue=("revenue", "mean"),
    total_quantity=("quantity", "sum"),
    transaction_count=("product", "size")
).round(2)

display(region_summary.sort_values("total_revenue", ascending=False))
```

```python
region_product = sales.groupby(["region", "product"]).agg(
    revenue=("revenue", "sum"),
    units_sold=("quantity", "sum")
).reset_index()

display(region_product)
```

```python
sales["region_total"] = sales.groupby("region")["revenue"].transform("sum")
sales["region_average"] = sales.groupby("region")["revenue"].transform("mean")

sales["share_of_region_revenue"] = (
    sales["revenue"] / sales["region_total"] * 100
).round(2)

display(sales)
```

```python
large_regions = sales.groupby("region").filter(
    lambda group: group["revenue"].sum() > 100000
)

display(large_regions.sort_values("revenue", ascending=False))
```

```python
sales["product_rank_in_region"] = (
    sales.groupby("region")["revenue"]
    .rank(method="dense", ascending=False)
)

display(
    sales.sort_values(["region", "product_rank_in_region"])
)
```

```python
pivot = pd.pivot_table(
    sales,
    index="region",
    columns="product",
    values="revenue",
    aggfunc="sum",
    fill_value=0,
    margins=True
)

display(pivot)
```

```python
best_region = region_summary["total_revenue"].idxmax()
best_product = sales.groupby("product")["revenue"].sum().idxmax()

print("Highest-revenue region:", best_region)
print("Highest-revenue product:", best_product)

region_summary.to_csv("regional_aggregation.csv")
```
