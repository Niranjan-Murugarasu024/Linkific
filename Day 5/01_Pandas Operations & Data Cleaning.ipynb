```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "employee_id": [101, 102, 103, 104, 105, 105],
    "name": [" Alice ", "BOB", "Charlie  ", None, "Eva", "Eva"],
    "department": ["IT", "HR", "IT", "Finance", "HR", "HR"],
    "salary": [75000, 52000, np.nan, 61000, 58000, 58000],
    "experience": [4, 2, 6, 3, np.nan, np.nan],
    "city": ["Chennai", "Coimbatore", "Chennai", None, "Madurai", "Madurai"]
})

display(df)
```

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print("Data types:")
print(df.dtypes)

display(df.describe(include="all").T)
```

```python
missing_report = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_percentage": (df.isna().mean() * 100).round(2)
})

display(missing_report)
```

```python
print("Duplicates before:", df.duplicated().sum())

df = df.drop_duplicates(subset=["employee_id"], keep="first")

print("Duplicates after:", df.duplicated(subset=["employee_id"]).sum())
display(df)
```

```python
df.columns = df.columns.str.strip().str.lower()

df["name"] = df["name"].astype("string").str.strip().str.title()
df["department"] = df["department"].astype("string").str.strip().str.title()
df["city"] = df["city"].astype("string").str.strip().str.title()

display(df)
```

```python
df["name"] = df["name"].fillna("Unknown")
df["department"] = df["department"].fillna("Unassigned")
df["city"] = df["city"].fillna("Unknown")

df["salary"] = df["salary"].fillna(
    df.groupby("department")["salary"].transform("median")
)

df["experience"] = df["experience"].fillna(
    df["experience"].median()
)

display(df)
print("Remaining missing values:")
print(df.isna().sum())
```

```python
df["employee_id"] = df["employee_id"].astype("int64")
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")
df["experience"] = pd.to_numeric(df["experience"], errors="coerce")

assert df["employee_id"].is_unique
assert df["salary"].notna().all()
assert df["salary"].ge(0).all()

print("Data quality checks passed.")
```

```python
df["annual_salary"] = df["salary"]
df["monthly_salary"] = (df["annual_salary"] / 12).round(2)

df["experience_level"] = pd.cut(
    df["experience"],
    bins=[-1, 2, 5, np.inf],
    labels=["Junior", "Mid-Level", "Senior"]
)

display(df)
```

```python
df = df.sort_values("salary", ascending=False)

display(df)

df.to_csv("cleaned_employees.csv", index=False)
print("Saved cleaned_employees.csv")
```
